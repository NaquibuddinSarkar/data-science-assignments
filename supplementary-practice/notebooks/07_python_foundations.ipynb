```python
# Iteration, filtering and comprehensions
values = [12, 7, 19, 4, 22, 9, 15]
above_ten = [value for value in values if value > 10]
squared = {value: value**2 for value in values}
print("Above ten:", above_ten)
print("Squares:", squared)
```

```python
# Functions with validation
def summarise_numbers(numbers):
    if not numbers:
        raise ValueError("numbers cannot be empty")
    if not all(isinstance(x, (int, float)) for x in numbers):
        raise TypeError("all values must be numeric")
    return {
        "count": len(numbers),
        "minimum": min(numbers),
        "maximum": max(numbers),
        "mean": sum(numbers) / len(numbers),
    }

summarise_numbers(values)
```

```python
# Exceptions: handle expected errors rather than hiding all exceptions
def safe_divide(numerator, denominator):
    try:
        return numerator / denominator
    except ZeroDivisionError:
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

```python
# A simple reusable transformation
records = [
    {"department": "Engineering", "salary": 65000},
    {"department": "Analytics", "salary": 59000},
    {"department": "Engineering", "salary": 72000},
    {"department": "Operations", "salary": 51000},
]
totals = {}
counts = {}
for record in records:
    department = record["department"]
    totals[department] = totals.get(department, 0) + record["salary"]
    counts[department] = counts.get(department, 0) + 1

average_salary = {key: totals[key] / counts[key] for key in totals}
average_salary
```
